```python
import pandas as pd

families = pd.read_csv("families.csv", sep=";")
couples = pd.read_csv("couples.csv", sep=";")
singles = pd.read_csv("Singles.csv", sep=";")
families
```

```python
# Combine into one dataset (rows are now indistinguishable)
combined = pd.concat([families, couples, singles], ignore_index=True)

# The first column is municipality name (currently called "Unnamed: 0")
combined = combined.rename(columns={"Unnamed: 0": "Municipality"})

# Group by municipality and take mean across the three datasets
income = combined.groupby("Municipality", as_index=False).mean(numeric_only=True)

income
```

```python
income = income[income["Municipality"] != "All Denmark"]
```

```python
income["Municipality"] = income["Municipality"].replace({
    "Vesthimmerlands": "Vesthimmerland",
    "Nordfyn": "Nordfyns"
})
```

```python
from collections import defaultdict

# --- Base mapping from official Danish regions ---
region_to_municipalities = {
    "Capital Region": [
        "Copenhagen", "Frederiksberg", "Gentofte", "Gladsaxe", "Helsingør",
        "Hillerød", "Hvidovre", "Lyngby-Taarbæk", "Rødovre", "Ballerup",
        "Brøndby", "Dragør", "Egedal", "Fredensborg", "Frederikssund",
        "Furesø", "Gribskov", "Halsnæs", "Herlev", "Høje-Taastrup",
        "Hørsholm", "Ishøj", "Rudersdal", "Tårnby", "Vallensbæk", 
        "Albertslund", "Allerød", "Glostrup", "Bornholm"
    ],
    "Zealand Region": [
        "Roskilde", "Holbæk", "Næstved", "Slagelse", "Køge",
        "Guldborgsund", "Vordingborg", "Kalundborg", "Ringsted",
        "Sorø", "Lejre", "Stevns", "Solrød", "Faxe", "Odsherred",
        "Greve", "Lolland", 
    ],
    "Central Jutland": [
        "Aarhus", "Randers", "Viborg", "Silkeborg", "Herning",
        "Horsens", "Skanderborg", "Holstebro", "Skive",
        "Ringkøbing-Skjern", "Favrskov", "Hedensted",
        "Syddjurs", "Ikast-Brande", "Norddjurs", "Struer",
        "Odder", "Lemvig", "Samsø"
    ],
    "North Jutland": [
        "Aalborg", "Hjørring", "Frederikshavn", "Thisted",
        "Mariagerfjord", "Jammerbugt", "Vesthimmerland", 
        "Brønderslev", "Rebild", "Morsø", "Læsø"
    ],
    "Southern Denmark": [
        "Odense", "Esbjerg", "Vejle", "Kolding", "Sønderborg",
        "Aabenraa", "Svendborg", "Haderslev", "Faaborg-Midtfyn",
        "Varde", "Fredericia", "Vejen", "Assens", "Tønder",
        "Middelfart", "Nyborg", "Nordfyns", "Billund",
        "Kerteminde", "Langeland", "Ærø", "Fanø"
    ]
}
# --- normalization helper (important for DK names) ---
def normalize(name):
    return (name.strip()
            .lower()
            .replace("å", "aa")
            .replace("ø", "oe")
            .replace("æ", "ae"))

# --- build reverse lookup ---
municipality_to_region = {}

for region, munis in region_to_municipalities.items():
    for m in munis:
        municipality_to_region[normalize(m)] = region

def get_region(muni):
    if muni == "All Denmark":
        return None  # or "Denmark Total"
    return municipality_to_region.get(normalize(muni), "Other islands")

income["Region"] = income["Municipality"].apply(get_region)

missing = income[
    income["Region"].isna() | (income["Region"] == "Other islands")
]["Municipality"].unique()

print("Unmatched municipalities:", missing)
```

```python
income_muni_long = income.melt(
    id_vars=["Municipality", "Region"],
    var_name="year",
    value_name="income"
)

income_muni_long["year"] = income_muni_long["year"].astype(int)

income_muni_long
```

```python
region_wide = income.groupby("Region").mean(numeric_only=True).reset_index()

income_region_long = region_wide.melt(
    id_vars=["Region"],
    var_name="year",
    value_name="income"
)

income_region_long["year"] = income_region_long["year"].astype(int)

income_region_long
```
